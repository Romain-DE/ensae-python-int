# Rappels

```python
import pandas
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
df_titanic = pandas.read_csv(url)
df_titanic.columns
```

```python
pandas.pivot_table(
    df_titanic, 
    index=["Sex", "Pclass"], 
    values="Survived")
```

```python
# Noter que le résultat précédent est équivalent à 
df_titanic.groupby(["Sex", "Pclass"])["Survived"].mean()
```

```python
pandas.pivot_table(df_titanic, 
                   index=["Sex", "Pclass"], 
                   columns="Survived", 
                   values="Fare")
```

```python
# Noter l'utilisation des margins
pandas.pivot_table(df_titanic, 
                   index=["Sex", "Pclass"], 
                   columns="Survived", 
                   values="Fare", margins = True)
```

```python
pandas.pivot_table(df_titanic, index=["Sex"], columns="Survived", values=["SibSp", "Parch"])
```

```python
# Crosstab
pandas.crosstab(df_titanic.Sex, df_titanic.Survived)
```
