```python
## Créer un dict avec une for comprehension à partir d'une liste
# le dictionnaire doit contenir comme clé les éléments de la liste suivante, et comme valeur la longueur de la string associée
intro_list = ["coucou", "bonjour", "hola", "salut", "hé", "hello"]

intro_dict = {key_name:len(key_name) for key_name in intro_list}
intro_dict
```

```python
## Modifier un dict à l'aide d'une for comprehension
# Modifier le dictionnaire obtenu précédemment pour ne garder que les clé dont la valeur est un nombre pair
{key:value for key, value in intro_dict.items() if value%2==0}
```

```python
# Parcourir les éléments d'un dict
# les clés
print("###Les clés###")
for key in intro_dict.keys():
    print(key)

# les valeurs
print("###Les valeurs###")
for value in intro_dict.values():
    print(value)

# les deux items en même temps
print("###Les clés et les valeurs###")
for key, value in intro_dict.items():
    print(key, value)
```

```python
# Bonus culture générale
# L'approche "Easier to ask for forgiveness than permission" (EAFP) consiste 
# à essayer d'accéder à une clé, et de gérer l'erreur si elle n'existe pas

try:
    print(f"la valeur associée à la clé 'citron' est {fruits['citron']}")
except KeyError:
    print("la clé 'citron' n'existe pas dans le dictionnaire")
# L'approche "Look before you leap" (LBYL) consiste à vérifier si la clé existe avant d'y accéder

if "citron" in fruits:
    print(f"la valeur associée à la clé 'citron' est {fruits['citron']}")
else:
    print("la clé 'citron' n'existe pas dans le dictionnaire")
```
