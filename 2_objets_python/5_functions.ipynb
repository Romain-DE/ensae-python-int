```python
from typing import Optional


def sum_int(a: int, b: int, c: Optional[int] = 0 ) -> int:
    if c :
        return a + b + c
    else : 
        return a + b
```

```python
def sum_int(**kwargs) -> int:
    count = 0
    for value in kwargs.values():
        count += value
    return count

sum_int(**{"a": 2, "b" : 3})
```

```python
sum_int(**{"a": 2, "b" : 3, "c": 4})
```

```python
def sum_int(my_dict : dict) -> int:
    count = 0
    for value in my_dict.values():
        count += value
    return count

sum_int({"a": 2, "b" : 3})
```

```python
# args et kwargs
# Dans l'exemple suivant, x et y sont dits "positional arguments" et z "keyword argument"
def g(x, y, z=1): 
    return (x + y) / z

# Les positional args sont paquetés dans un tuple, et les kwargs dans un dict. 
# Il est donc possible d'apppeler g de la façon suivante :
pos_args = (1, 2)
kwargs = {"z" : 5}
g(*pos_args, **kwargs)
# où * permet de déployer un tuple
# et ** un dict
```
