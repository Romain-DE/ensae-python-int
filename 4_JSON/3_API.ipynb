### API Météo
Ou comment paramétrer un appel HTTP

```python
import json
url = "https://api.open-meteo.com/v1/forecast"
params = {
    "latitude": 48.8566,
    "longitude": 2.3522,
    "hourly": "temperature_2m",
    "start_hour": "2025-04-01T00:00",
    "end_hour": "2025-04-01T23:00",
    "timezone": "Europe/Paris"
}
res = requests.get(url, params=params)
weather = res.json()

# A quoi ressemble la réponse ?
print(json.dumps(weather, indent=2))
```

```python
weather["hourly"]
```
